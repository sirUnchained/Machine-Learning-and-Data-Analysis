```python
# ! wget https://archive.ics.uci.edu/static/public/468/online+shoppers+purchasing+intention+dataset.zip -o online_shoppers_intention.csv
```

The dataset contains 12,330 sessions and 17 features (variables), with the goal of predicting whether a session ends with a purchase (the `Revenue` attribute). It's a multivariate dataset suitable for classification and clustering tasks, **with no missing values**.

| Variable Name | Type | Description & Business Insight |
| :--- | :--- | :--- |
| `Administrative` | Integer | Number of pages visited by the user in the "Administrative" category (e.g., account, contact, terms). A higher count might indicate account management, not necessarily purchase intent. |
| `Administrative_Duration` | Float | Total time (in seconds) spent on administrative pages. Long durations could mean confusion or detailed reading. |
| `Informational` | Integer | Number of pages visited in the "Informational" category (e.g., about us, FAQs, blogs). This often indicates research and information-seeking behavior. |
| `Informational_Duration` | Float | Total time (in seconds) spent on informational pages. |
| `ProductRelated` | Integer | Number of pages visited related to products (e.g., product pages, categories). This is a **strong indicator of purchase intent**; higher values are typically positive. |
| `ProductRelated_Duration` | Float | Total time (in seconds) spent on product-related pages. This is also a **key indicator**; spending more time here usually suggests higher interest. |
| `BounceRates` | Continuous | The average bounce rate of the pages visited by the user. A "bounce" is when a user enters and leaves without triggering any other server requests. High bounce rates on pages visited can be a negative signal. |
| `ExitRates` | Continuous | The average exit rate of the pages visited. This is the percentage of views of a page that were the last in the session. High exit rates on product pages, for example, might indicate a problem. |
| `PageValues` | Continuous | The average value of the pages visited, representing how much a page contributes to an eventual transaction. This is a **powerful predictor**, as pages with higher historical value are more likely to lead to a purchase. |
| `SpecialDay` | Integer | Closeness of the visit date to a special day (e.g., Valentine's Day, Mother's Day). The value is non-zero only in the days leading up to the event (e.g., 0.4, 0.6, 1.0), indicating higher promotional intent. |
| `Weekend` | Boolean (Categorical) | Indicates whether the visit occurred on a weekend (True) or not (False). Shopping behavior often differs on weekends. |
| `Month` | Categorical | The month of the year (e.g., Jan, Feb, Mar, ...). Useful for capturing seasonal trends and marketing campaign effects. |
| `VisitorType` | Categorical | Whether the visitor is a "New Visitor", "Returning Visitor", or "Other". Returning visitors are often more likely to purchase. |
| `OperatingSystems` | Integer | The operating system used by the visitor (coded as integers, e.g., 1, 2, 3). Can be used to segment users. |
| `Browser` | Integer | The browser used by the visitor (coded as integers). |
| `Region` | Integer | The geographic region of the visitor (coded as integers). |
| `TrafficType` | Integer | The traffic source (e.g., 1 for direct, 2 for search engines). Useful for understanding which marketing channels perform best. |
| `Revenue (Target)` | Boolean | Indicates whether the session resulted in a transaction (True) or not (False). This is the class label for prediction tasks. The dataset is imbalanced: 84.5% of sessions (False) did **not** end in a purchase, and 15.5% (True) did. |

## Preprocess data

### Remove `SpecialDay` and `BounceRates`

```python
df = df.drop(columns=['BounceRates', 'SpecialDay'])

df_cat = df.select_dtypes(exclude="float64")
df_num = df.select_dtypes(include="float64")
```

### Encode label VisitorType

```python
le = LabelEncoder()
df['VisitorType'] = le.fit_transform(df['VisitorType'])
```

### Remove outlier datas

```python
def remove_outlier(df: pd.DataFrame, col_name: str):
    limit = 2.5
    q1 = df[col_name].quantile(0.25)
    q3 = df[col_name].quantile(0.75)
    iqr = q3 - q1
    df = df[df[col_name] >= q1 - limit * iqr]
    df = df[df[col_name] <= q3 + limit * iqr]
    return df
```

```python
print(df.shape)
for n in df_num.columns:
    df = remove_outlier(df, col_name=n)
print(df.shape)
```

### Our dataset shall have 0|1 not booleans

```python
df['Revenue'] = df['Revenue'].replace(to_replace=[True, False], value=[1, 0])
df['Weekend'] = df['Weekend'].replace(to_replace=[True, False], value=[1, 0])
```

### Make Dataset balanced

```python
X = df.drop(columns=['Revenue'])
y = df['Revenue']
```

```python
smote = SMOTE(random_state=RANDOM_SEED)
X_resample, y_resample = smote.fit_resample(X, y)
```

```python
y_resample.hist()
```

### Train test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_resample, y_resample, test_size=0.2, random_state=RANDOM_SEED)
```

## model

```python
xgb_model = xgb.XGBClassifier(
    objective='binary:logistic',
    learning_rate=0.1,
    n_estimators=200,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=RANDOM_SEED,
    eval_metric='logloss'  
)

xgb_model.fit(X_train, y_train, eval_set=[(X_train, y_train), (X_test, y_test)])
y_pred = xgb_model.predict(X_test)
print(classification_report(y_test, y_pred))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
rfc = RandomForestClassifier(random_state=RANDOM_SEED)
rfc.fit(X_train, y_train)

y_pred = rfc.predict(X_test)
print(classification_report(y_test, y_pred))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
logreg = LogisticRegression(random_state=RANDOM_SEED, max_iter=5000)
logreg.fit(X_train, y_train)

y_pred = logreg.predict(X_test)
print(classification_report(y_test, y_pred))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
dtc = DecisionTreeClassifier(random_state=RANDOM_SEED)
dtc_model_VisitorType = dtc.fit(X_train, y_train)

y_pred = dtc.predict(X_test)
print(classification_report(y_test, y_pred))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
svc = SVC(random_state=RANDOM_SEED, kernel="rbf")
svc_model = svc.fit(X_train, y_train)
y_pred = svc_model.predict(X_test)
print(classification_report(y_test, y_pred))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```
